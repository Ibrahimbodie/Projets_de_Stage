# 01 — Exploration du Benchmark
**RQ** : —

Analyse descriptive du jeu de données : distribution des niveaux, catégories, types de questions.

```python
import sys, os, pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
sys.path.insert(0, os.path.abspath(".."))
sys.path.insert(0, os.path.abspath("../src"))

from notebooks.lib.reporter import load_benchmark
from notebooks.lib.plotter import histogram, barplot, set_style

df = load_benchmark()
print(f"Benchmark chargé : {len(df)} questions")
print(f"Colonnes : {list(df.columns)}")
df.head(3)
```

```python
# Distribution par niveau de complexité
print("=== Distribution par Niveau ===")
print(df['Niveau_Complexite'].value_counts())
print()
print("=== Distribution par Type de Réponse ===")
print(df['Type_Reponse'].value_counts())
print()
print("=== Distribution par Catégorie Métier ===")
print(df['Categorie_Metier'].value_counts())
```

```python
# Visualisations
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

df['Niveau_Complexite'].value_counts().plot(kind='bar', ax=axes[0], color='#1F384B')
axes[0].set_title('Questions par Niveau de Complexité')
axes[0].tick_params(axis='x', rotation=45)

df['Type_Reponse'].value_counts().plot(kind='bar', ax=axes[1], color='#2E86AB')
axes[1].set_title('Questions par Type de Réponse')
axes[1].tick_params(axis='x', rotation=45)

df['Categorie_Metier'].value_counts().plot(kind='bar', ax=axes[2], color='#A23B72')
axes[2].set_title('Questions par Catégorie Métier')
axes[2].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.savefig('../outputs/figures/histograms/benchmark_distributions.png', dpi=200, bbox_inches='tight')
plt.show()
```

```python
# Analyse des longueurs de questions
df['question_len'] = df['Question'].apply(len)
print(f"Longueur des questions : min={df['question_len'].min()}, max={df['question_len'].max()}, moy={df['question_len'].mean():.0f}")

fig, ax = plt.subplots(figsize=(10, 4))
sns.histplot(df['question_len'], bins=15, color='#1F384B', edgecolor='white', ax=ax)
ax.set_title('Distribution de la longueur des questions')
ax.set_xlabel('Nombre de caractères')
plt.savefig('../outputs/figures/histograms/question_length.png', dpi=200, bbox_inches='tight')
plt.show()
```

## Synthèse
- Benchmark : 30 questions
- 4 niveaux de complexité (8 extraction, 8 synthèse, 7 inférence, 7 hors-domaine)
- 3 types de réponses (factuel, procédural, conditionnel)
- 6 catégories métier (finance, général, scolarité, visa, résidence, discipline)
